```python
import numpy as np
```

```python
vector1=np.array([1,2,3])
vector2=np.array([4,5,6])
print("Vector 1 = ",vector1)
print("Vector 2 = ",vector2)
```

```python
addition=vector1+vector2
print("Vector Addition = ",addition)
```

```python
multiply=np.matmul(vector1,vector2)
print("Multiplication vector 1 and 2 =",multiply)
```

```python
A=np.array([[1,2],[3,4]])
B=np.array([[5,6],[7,8]])
print("Matrix A : ")
print(A)
print("Matrix B : ")
print(B)
```

```python
C=A@B
print(C)
```
