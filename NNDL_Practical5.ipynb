```python
# ================================================================
# FLOWER IMAGE CLASSIFICATION USING TRANSFER LEARNING
# PRE-TRAINED MODEL: MobileNetV2
# ================================================================


# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import os
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf

from tensorflow.keras import layers, models
from google.colab import files
from PIL import Image

print("================================================")
print("FLOWER CLASSIFICATION USING MOBILENETV2")
print("================================================")

print("TensorFlow Version:", tf.__version__)


# ================================================================
# 2. DOWNLOAD FLOWER DATASET
# ================================================================

dataset_url = "https://storage.googleapis.com/download.tensorflow.org/example_images/flower_photos.tgz"

print("\nDownloading flower dataset...")

data_dir = tf.keras.utils.get_file(
    "flower_photos",
    origin=dataset_url,
    untar=True
)

# ------------------------------------------------
# IMPORTANT:
# TensorFlow may extract the dataset as:
#
# flower_photos/
#       flower_photos/
#             daisy/
#             dandelion/
#             roses/
#             sunflowers/
#             tulips/
#
# So we check for the inner folder.
# ------------------------------------------------

inner_folder = os.path.join(data_dir, "flower_photos")

if os.path.isdir(inner_folder):
    data_dir = inner_folder

print("\nDataset path:")
print(data_dir)


# ================================================================
# 3. CHECK DATASET FOLDERS
# ================================================================

print("\n================================================")
print("CHECKING DATASET")
print("================================================")

folders = sorted([
    folder for folder in os.listdir(data_dir)
    if os.path.isdir(os.path.join(data_dir, folder))
])

print("\nFlower folders found:")

for folder in folders:
    print(" -", folder)

print("\nNumber of flower categories:", len(folders))


# ================================================================
# 4. IMAGE PARAMETERS
# ================================================================

IMG_SIZE = 224
BATCH_SIZE = 32
SEED = 123


# ================================================================
# 5. LOAD TRAINING DATASET
# ================================================================

print("\n================================================")
print("LOADING TRAINING DATA")
print("================================================")

train_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.20,
    subset="training",
    seed=SEED,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)


# ================================================================
# 6. LOAD VALIDATION DATASET
# ================================================================

print("\n================================================")
print("LOADING VALIDATION DATA")
print("================================================")

val_ds = tf.keras.utils.image_dataset_from_directory(
    data_dir,
    validation_split=0.20,
    subset="validation",
    seed=SEED,
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE
)


# ================================================================
# 7. GET FLOWER CLASS NAMES
# ================================================================

class_names = train_ds.class_names

print("\n================================================")
print("FLOWER CATEGORIES")
print("================================================")

for i, name in enumerate(class_names):
    print(f"{i} : {name}")

print("\nTotal number of classes:", len(class_names))


# ================================================================
# 8. VERIFY THAT THERE ARE MULTIPLE CLASSES
# ================================================================

if len(class_names) < 2:
    raise ValueError(
        "\nERROR: Only one class was detected!\n"
        "Check the dataset directory structure."
    )

print("\nDataset correctly contains multiple flower categories.")


# ================================================================
# 9. DISPLAY SAMPLE IMAGES
# ================================================================

print("\nDisplaying sample flower images...")

plt.figure(figsize=(12, 8))

for images, labels in train_ds.take(1):

    for i in range(min(9, len(images))):

        plt.subplot(3, 3, i + 1)

        plt.imshow(images[i].numpy().astype("uint8"))

        plt.title(class_names[labels[i]])

        plt.axis("off")

plt.tight_layout()
plt.show()


# ================================================================
# 10. OPTIMIZE DATASET PERFORMANCE
# ================================================================

AUTOTUNE = tf.data.AUTOTUNE

train_ds = train_ds.prefetch(
    buffer_size=AUTOTUNE
)

val_ds = val_ds.prefetch(
    buffer_size=AUTOTUNE
)


# ================================================================
# 11. DATA AUGMENTATION
# ================================================================

data_augmentation = tf.keras.Sequential([

    layers.RandomFlip(
        "horizontal"
    ),

    layers.RandomRotation(
        0.1
    ),

    layers.RandomZoom(
        0.1
    )

], name="data_augmentation")


# ================================================================
# 12. LOAD PRE-TRAINED MOBILENETV2
# ================================================================

print("\n================================================")
print("LOADING PRE-TRAINED MOBILENETV2")
print("================================================")

base_model = tf.keras.applications.MobileNetV2(

    input_shape=(
        IMG_SIZE,
        IMG_SIZE,
        3
    ),

    include_top=False,

    weights="imagenet"
)


# ================================================================
# 13. FREEZE PRE-TRAINED LAYERS
# ================================================================

base_model.trainable = False

print("MobileNetV2 loaded successfully.")
print("Pre-trained convolutional layers are frozen.")


# ================================================================
# 14. MOBILENETV2 PREPROCESSING
# ================================================================

preprocess_input = (
    tf.keras.applications.mobilenet_v2.preprocess_input
)


# ================================================================
# 15. BUILD TRANSFER LEARNING MODEL
# ================================================================

print("\n================================================")
print("BUILDING TRANSFER LEARNING MODEL")
print("================================================")

inputs = tf.keras.Input(
    shape=(
        IMG_SIZE,
        IMG_SIZE,
        3
    )
)


# Data augmentation
x = data_augmentation(inputs)


# MobileNetV2 preprocessing
x = preprocess_input(x)


# Pre-trained MobileNetV2
x = base_model(
    x,
    training=False
)


# Global Average Pooling
x = layers.GlobalAveragePooling2D()(x)


# Dropout
x = layers.Dropout(
    0.2
)(x)


# Classification layer
outputs = layers.Dense(
    len(class_names),
    activation="softmax"
)(x)


# Create final model
model = models.Model(
    inputs,
    outputs
)


# ================================================================
# 16. COMPILE MODEL
# ================================================================

model.compile(

    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0001
    ),

    loss="sparse_categorical_crossentropy",

    metrics=[
        "accuracy"
    ]
)


# ================================================================
# 17. CHECK MODEL OUTPUT
# ================================================================

print("\n================================================")
print("MODEL INFORMATION")
print("================================================")

print(
    "Number of output classes:",
    len(class_names)
)

print(
    "Output classes:",
    class_names
)

model.summary()


# ================================================================
# 18. TRAIN MODEL
# ================================================================

print("\n================================================")
print("TRAINING MODEL")
print("================================================")

EPOCHS = 10

history = model.fit(

    train_ds,

    validation_data=val_ds,

    epochs=EPOCHS
)


# ================================================================
# 19. EVALUATE MODEL
# ================================================================

print("\n================================================")
print("MODEL EVALUATION")
print("================================================")

loss, accuracy = model.evaluate(
    val_ds,
    verbose=1
)

print(
    "\nValidation Loss:",
    round(loss, 4)
)

print(
    "Validation Accuracy:",
    round(accuracy * 100, 2),
    "%"
)


# ================================================================
# 20. PLOT TRAINING ACCURACY AND VALIDATION ACCURACY
# ================================================================

acc = history.history["accuracy"]

val_acc = history.history["val_accuracy"]

loss_values = history.history["loss"]

val_loss = history.history["val_loss"]

epochs_range = range(
    1,
    EPOCHS + 1
)


plt.figure(
    figsize=(14, 5)
)


# -----------------------------
# Accuracy
# -----------------------------

plt.subplot(
    1,
    2,
    1
)

plt.plot(
    epochs_range,
    acc,
    marker="o",
    label="Training Accuracy"
)

plt.plot(
    epochs_range,
    val_acc,
    marker="o",
    label="Validation Accuracy"
)

plt.xlabel("Epoch")

plt.ylabel("Accuracy")

plt.title(
    "Training and Validation Accuracy"
)

plt.legend()

plt.grid(True)


# -----------------------------
# Loss
# -----------------------------

plt.subplot(
    1,
    2,
    2
)

plt.plot(
    epochs_range,
    loss_values,
    marker="o",
    label="Training Loss"
)

plt.plot(
    epochs_range,
    val_loss,
    marker="o",
    label="Validation Loss"
)

plt.xlabel("Epoch")

plt.ylabel("Loss")

plt.title(
    "Training and Validation Loss"
)

plt.legend()

plt.grid(True)


plt.tight_layout()

plt.show()


# ================================================================
# 21. SAVE TRAINED MODEL
# ================================================================

model_path = "flower_mobilenetv2_model.keras"

model.save(
    model_path
)

print("\n================================================")
print("MODEL SAVED")
print("================================================")

print(
    "Saved as:",
    model_path
)


# ================================================================
# 22. UPLOAD NEW FLOWER IMAGE
# ================================================================

print("\n")
print("================================================")
print("UPLOAD A FLOWER IMAGE")
print("================================================")

print(
    "Click 'Choose Files' and select a flower image."
)

uploaded = files.upload()


# ================================================================
# 23. PREDICT UPLOADED IMAGE
# ================================================================

for filename in uploaded.keys():

    print("\n")
    print("Processing image:", filename)


    # ------------------------------------------------
    # Open image
    # ------------------------------------------------

    image = Image.open(
        filename
    ).convert("RGB")


    # ------------------------------------------------
    # Display uploaded image
    # ------------------------------------------------

    plt.figure(
        figsize=(7, 7)
    )

    plt.imshow(
        image
    )

    plt.axis(
        "off"
    )

    plt.title(
        "Uploaded Flower Image"
    )

    plt.show()


    # ------------------------------------------------
    # Resize image
    # ------------------------------------------------

    image_resized = image.resize(
        (
            IMG_SIZE,
            IMG_SIZE
        )
    )


    # ------------------------------------------------
    # Convert image to NumPy array
    # ------------------------------------------------

    img_array = np.array(
        image_resized
    ).astype(
        np.float32
    )


    # ------------------------------------------------
    # Add batch dimension
    # ------------------------------------------------

    img_array = np.expand_dims(
        img_array,
        axis=0
    )


    # ------------------------------------------------
    # Make prediction
    # ------------------------------------------------

    predictions = model.predict(
        img_array,
        verbose=0
    )[0]


    # ------------------------------------------------
    # Find predicted class
    # ------------------------------------------------

    predicted_index = np.argmax(
        predictions
    )

    predicted_class = class_names[
        predicted_index
    ]

    confidence = (
        predictions[predicted_index]
        * 100
    )


    # =================================================
    # 24. DISPLAY PREDICTION RESULT
    # =================================================

    print("\n")
    print("================================================")
    print("              FLOWER PREDICTION")
    print("================================================")

    print(
        "Uploaded Image   :",
        filename
    )

    print(
        "Predicted Flower :",
        predicted_class
    )

    print(
        "Confidence       : {:.2f}%".format(
            confidence
        )
    )

    print("================================================")


    # =================================================
    # 25. DISPLAY ALL CLASS PROBABILITIES
    # =================================================

    print("\nPrediction probabilities:")

    for i, flower in enumerate(
        class_names
    ):

        probability = (
            predictions[i]
            * 100
        )

        print(
            "{:<15} : {:.2f}%".format(
                flower,
                probability
            )
        )


    # =================================================
    # 26. DISPLAY PROBABILITY BAR GRAPH
    # =================================================

    plt.figure(
        figsize=(10, 5)
    )

    probabilities = (
        predictions * 100
    )

    bars = plt.bar(
        class_names,
        probabilities,
        color="green"
    )

    plt.xlabel(
        "Flower Category"
    )

    plt.ylabel(
        "Probability (%)"
    )

    plt.title(
        "Flower Classification Probabilities"
    )

    plt.ylim(
        0,
        100
    )

    plt.xticks(
        rotation=20
    )


    # Add percentage above bars
    for bar, probability in zip(
        bars,
        probabilities
    ):

        plt.text(
            bar.get_x()
            + bar.get_width() / 2,

            bar.get_height()
            + 1,

            "{:.2f}%".format(
                probability
            ),

            ha="center"
        )


    plt.tight_layout()

    plt.show()


# ================================================================
# END OF PROGRAM
# ================================================================

print("\n")
print("================================================")
print("PROGRAM COMPLETED")
print("================================================")
print("MobileNetV2 Transfer Learning completed.")
print("You can upload another flower image by")
print("running the upload/prediction section again.")
print("================================================")
```
